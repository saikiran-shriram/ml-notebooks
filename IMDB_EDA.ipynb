```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("IMDB Dataset.csv")
print(df.shape)
print(df.head(5))
print(df['sentiment'].value_counts())
print(df['review'].str.len().describe())
print(df.isnull().sum())
print(df[df['sentiment']=='negative']['review'].iloc[0][:500])
print(df[df['sentiment']=='positive']['review'].iloc[0][:500])
```

```python
import re

def clean_text(text):
    text = re.sub('<.*?>', '', text)  # remove HTML tags like <br />
    text = text.lower()              # lowercase
    return text

df['clean_review'] = df['review'].apply(clean_text)
print(df['clean_review'].iloc[0][:300])
```
