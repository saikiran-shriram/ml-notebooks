```python
#importing libraries
import matplotlib.pyplot as plt
import numpy as np

#creating variables
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

w = 0
b = 0

#predict function
def predict (x,w,b):
  return w*x + b

#cost function
def cost (x,y,w,b):
  m = len(x)
  predictions = predict(x,w,b)
  cost = (1/(2*m)) * np.sum((predictions - y)**2)
  return cost

#gradient descent function
def gradient_descent(x,y,w,b,learning_rate,iterations):
  m = len(x)
  for i in range(iterations):
    predictions = predict(x,w,b)
    dw = (1/m) * np.sum(x * (predictions - y))
    db = (1/m) * np.sum(predictions - y)
    w = w - learning_rate * dw
    b = b - learning_rate * db
  return w,b

w ,b = gradient_descent(x,y,w,b,0.01,1000)
print(w,b)

#graph
plt.scatter(x,y)
plt.plot(x,predict(x,w,b))
plt.title('Linear Regression')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
